```python
categorical_columns = [
    'Gender',
    'Category',
    'Season',
    'Size',
    'Subscription Status',
    'Discount Applied',
    'Payment Method',
    'Frequency of Purchases'
]

for column in categorical_columns:
    print("\n", column)
    print(df[column].value_counts())
```

```python
numerical_columns = [
    'Age',
    'Purchase Amount (USD)',
    'Review Rating',
    'Previous Purchases'
]

df[numerical_columns].describe()
```

```python
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[0, 18, 25, 35, 50, 65, 100],
    labels=[
        'Under 18',
        '18-25',
        '26-35',
        '36-50',
        '51-65',
        '65+'
    ]
)

df[['Age', 'Age Group']].head(10)
```

```python
df['Spending Segment'] = pd.cut(
    df['Purchase Amount (USD)'],
    bins=[0, 40, 70, 100],
    labels=[
        'Low Spender',
        'Medium Spender',
        'High Spender'
    ]
)

df[['Purchase Amount (USD)', 'Spending Segment']].head(10)
```

```python
category_spending = (
    df.groupby('Category')['Purchase Amount (USD)']
    .sum()
    .sort_values(ascending=False)
)

category_spending
```

```python
total_customers = df['Customer ID'].nunique()

print("Total Customers:", total_customers)
```

```python
total_revenue = df['Purchase Amount (USD)'].sum()

print("Total Purchase Value: $", total_revenue)
```

```python
average_purchase = df['Purchase Amount (USD)'].mean()

print("Average Purchase Amount: $", round(average_purchase, 2))
```

```python
average_rating = df['Review Rating'].mean()

print("Average Review Rating:", round(average_rating, 2))
```

```python
subscription_rate = (
    df['Subscription Status'].eq('Yes').mean() * 100
)

print("Subscription Rate:", round(subscription_rate, 2), "%")
```

```python
discount_rate = (
    df['Discount Applied'].eq('Yes').mean() * 100
)

print("Discount Usage Rate:", round(discount_rate, 2), "%")
```

```python
kpi = pd.DataFrame({
    'KPI': [
        'Total Customers',
        'Total Purchase Value',
        'Average Purchase',
        'Average Rating',
        'Subscription Rate',
        'Discount Usage Rate'
    ],
    'Value': [
        total_customers,
        total_revenue,
        round(average_purchase, 2),
        round(average_rating, 2),
        round(subscription_rate, 2),
        round(discount_rate, 2)
    ]
})

kpi
```
